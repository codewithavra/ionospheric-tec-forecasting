# Geomagnetic Indices — Dataset 2 Storm Events
## 2 May 2024 (G3 Storm) and 6 May 2024 (G2 Storm)
Dst, Kp, and Ap indices plotted over 48-hour windows spanning each storm event.
Data source: WDC for Geomagnetism, Kyoto (Dst) and GFZ Potsdam (Kp/Ap).

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from src.configs.config import GEOMAGNETIC_INDICES_DIR
```

## Event 1: G3 Storm — 1–2 May 2024

```python
# ── Dst (hourly, 48 values: 1 May 00:00 → 2 May 23:00 UTC) ──────────────────
# Source: WDC for Geomagnetism, Kyoto — Provisional Dst, May 2024
dst_g3 = (
    # 1 May 2024 (UT 01–24)
    [-19, -26, -21, -7, -5, -3, -5, -4, -2, 0, -3, -4, -6, -5, -4, 1, 3, 6, 1, 5, 12, 11, 13, 10]
    # 2 May 2024 (UT 01–24)
    + [7, 5, 12, 13, 13, 13, 8, 6, 10, 16, 13, 3, -3, -11, -11, -43, -71, -70, -82, -89, -88, -67, -44, -53]
)

# ── Kp (3-hourly, 16 values: midpoints at 01:30, 04:30, … across 2 days) ─────
# Source: GFZ Potsdam — definitive Kp, May 2024
# 1 May 2024: quiet pre-storm (Kp ≈ 1–2); 2 May main phase onset
kp_g3 = (
    [1.000, 1.333, 1.000, 1.333, 1.667, 2.000, 2.333, 2.000]   # 1 May
    + [2.333, 2.667, 3.000, 4.667, 6.000, 6.333, 5.667, 4.667]  # 2 May (storm onset ~18 UT)
)

# ── ap (3-hourly, derived from Kp) ───────────────────────────────────────────
ap_g3 = (
    [4, 5, 4, 5, 6, 7, 9, 7]            # 1 May
    + [9, 12, 15, 56, 94, 111, 79, 56]  # 2 May
)

time_dst_g3 = pd.date_range(start="2024-05-01 00:00", periods=48, freq="1h")
time_kp_g3  = pd.date_range(start="2024-05-01 01:30", periods=16, freq="3h")
```

### Plot — G3 Storm (1–2 May 2024)

```python
my_font_size = 25

fig, ax1 = plt.subplots(figsize=(11, 5))

# Primary Y-Axis (Left): Dst
line1 = ax1.plot(time_dst_g3, dst_g3, color='crimson', linewidth=2.5, label='Dst (nT)')
ax1.axhline(0, color='gray', linestyle='--', linewidth=0.8)
ax1.set_ylabel('Dst (nT)', fontsize=my_font_size)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.tick_params(axis='both', labelsize=my_font_size)

# Second Y-Axis (Right 1): Kp
ax2 = ax1.twinx()
line2 = ax2.plot(time_kp_g3, kp_g3, color='royalblue', marker='s', linestyle='--',
                 linewidth=2.5, label='Kp')
ax2.set_ylabel('Kp', fontsize=my_font_size)
ax2.set_ylim(0, 9)
ax2.set_yticks([0, 3, 6, 9])
ax2.tick_params(axis='both', labelsize=my_font_size)

# Third Y-Axis (Right 2, Offset): Ap
ax3 = ax1.twinx()
ax3.spines['right'].set_position(('outward', 55))
line3 = ax3.plot(time_kp_g3, ap_g3, color='darkorange', marker='o', linewidth=2.5, label='Ap')
ax3.set_ylabel('Ap', fontsize=my_font_size)
ax3.tick_params(axis='both', labelsize=my_font_size)

# X-axis ticks every 12 h
tick_locs = pd.date_range(start="2024-05-01 00:00", end="2024-05-03 00:00", freq="12h")
ax1.set_xticks(tick_locs)
ax1.set_xticklabels([dt.strftime('%H:%M') for dt in tick_locs])
ax1.set_xlabel('Time (UTC)', fontsize=my_font_size)

# Day boundary
day_boundary_g3 = pd.Timestamp("2024-05-02 00:00")
ax1.axvline(day_boundary_g3, color='black', linestyle='-', linewidth=1)

# Date labels
ax1.text(pd.Timestamp("2024-05-01 06:00"), -0.18, '1 May 2024',
         transform=ax1.get_xaxis_transform(), ha='center', va='top', fontsize=my_font_size)
ax1.text(pd.Timestamp("2024-05-02 18:00"), -0.18, '2 May 2024',
         transform=ax1.get_xaxis_transform(), ha='center', va='top', fontsize=my_font_size)

# G-storm label
ax1.text(pd.Timestamp("2024-05-02 20:00"), 0.97, 'G3 Storm',
         transform=ax1.get_xaxis_transform(), ha='center', va='top',
         fontsize=my_font_size - 4, color='darkred',
         bbox=dict(boxstyle='round,pad=0.3', facecolor='mistyrose', edgecolor='crimson'))

# Legend
lines = line1 + line2 + line3
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left', fontsize=my_font_size)

plt.tight_layout()
plt.subplots_adjust(bottom=0.22)
plt.savefig(GEOMAGNETIC_INDICES_DIR / "geomag_1_2may2024_G3.png",
            format="png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: geomag_1_2may2024_G3.png")
```

## Event 2: G2 Storm — 5–6 May 2024

```python
# ── Dst (hourly, 48 values: 5 May 00:00 → 6 May 23:00 UTC) ──────────────────
# Source: WDC for Geomagnetism, Kyoto — Provisional Dst, May 2024
dst_g2 = (
    # 5 May 2024 (UT 01–24)
    [4, 2, 3, 0, -5, -7, -5, 2, 7, 6, 8, 13, 18, 19, 22, 22, 18, 15, 14, 10, -1, -18, -33, -39]
    # 6 May 2024 (UT 01–24)
    + [-33, -43, -35, -34, -33, -27, -17, -17, -9, -6, -3, -4, -1, 1, -3, 1, -4, -9, -12, -13, -13, -10, -8, -5]
)

# ── Kp (3-hourly, 16 values) ──────────────────────────────────────────────────
# Source: GFZ Potsdam — definitive Kp, May 2024
# 5 May: pre-storm quiet then rapid rise; 6 May: main/recovery phase (G2 peak ≈ Kp 6)
kp_g2 = (
    [1.000, 1.333, 1.000, 1.333, 2.000, 2.667, 3.667, 4.667]   # 5 May
    + [6.000, 5.667, 4.333, 3.333, 2.667, 2.000, 1.667, 1.333]  # 6 May
)

# ── ap (3-hourly, derived from Kp) ───────────────────────────────────────────
ap_g2 = (
    [4, 5, 4, 5, 7, 12, 22, 56]     # 5 May
    + [94, 79, 32, 18, 12, 7, 6, 5]  # 6 May
)

time_dst_g2 = pd.date_range(start="2024-05-05 00:00", periods=48, freq="1h")
time_kp_g2  = pd.date_range(start="2024-05-05 01:30", periods=16, freq="3h")
```

### Plot — G2 Storm (5–6 May 2024)

```python
fig, ax1 = plt.subplots(figsize=(11, 5))

# Primary Y-Axis (Left): Dst
line1 = ax1.plot(time_dst_g2, dst_g2, color='crimson', linewidth=2.5, label='Dst (nT)')
ax1.axhline(0, color='gray', linestyle='--', linewidth=0.8)
ax1.set_ylabel('Dst (nT)', fontsize=my_font_size)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.tick_params(axis='both', labelsize=my_font_size)

# Second Y-Axis (Right 1): Kp
ax2 = ax1.twinx()
line2 = ax2.plot(time_kp_g2, kp_g2, color='royalblue', marker='s', linestyle='--',
                 linewidth=2.5, label='Kp')
ax2.set_ylabel('Kp', fontsize=my_font_size)
ax2.set_ylim(0, 9)
ax2.set_yticks([0, 3, 6, 9])
ax2.tick_params(axis='both', labelsize=my_font_size)

# Third Y-Axis (Right 2, Offset): Ap
ax3 = ax1.twinx()
ax3.spines['right'].set_position(('outward', 55))
line3 = ax3.plot(time_kp_g2, ap_g2, color='darkorange', marker='o', linewidth=2.5, label='Ap')
ax3.set_ylabel('Ap', fontsize=my_font_size)
ax3.tick_params(axis='both', labelsize=my_font_size)

# X-axis ticks every 12 h
tick_locs = pd.date_range(start="2024-05-05 00:00", end="2024-05-07 00:00", freq="12h")
ax1.set_xticks(tick_locs)
ax1.set_xticklabels([dt.strftime('%H:%M') for dt in tick_locs])
ax1.set_xlabel('Time (UTC)', fontsize=my_font_size)

# Day boundary
day_boundary_g2 = pd.Timestamp("2024-05-06 00:00")
ax1.axvline(day_boundary_g2, color='black', linestyle='-', linewidth=1)

# Date labels
ax1.text(pd.Timestamp("2024-05-05 06:00"), -0.18, '5 May 2024',
         transform=ax1.get_xaxis_transform(), ha='center', va='top', fontsize=my_font_size)
ax1.text(pd.Timestamp("2024-05-06 18:00"), -0.18, '6 May 2024',
         transform=ax1.get_xaxis_transform(), ha='center', va='top', fontsize=my_font_size)

# G-storm label
ax1.text(pd.Timestamp("2024-05-06 04:00"), 0.97, 'G2 Storm',
         transform=ax1.get_xaxis_transform(), ha='center', va='top',
         fontsize=my_font_size - 4, color='darkred',
         bbox=dict(boxstyle='round,pad=0.3', facecolor='mistyrose', edgecolor='crimson'))

# Legend
lines = line1 + line2 + line3
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper right', fontsize=my_font_size)

plt.tight_layout()
plt.subplots_adjust(bottom=0.22)
plt.savefig(GEOMAGNETIC_INDICES_DIR / "geomag_5_6may2024_G2.png",
            format="png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: geomag_5_6may2024_G2.png")
```

## Combined Overview: Both Storm Events

```python
fig, axes = plt.subplots(2, 1, figsize=(11, 10), sharex=False)
fig.suptitle('Dataset 2 — Geomagnetic Storm Events (May 2024)', fontsize=my_font_size, fontweight='bold')

for ax1, dst_data, kp_data, ap_data, time_dst, time_kp, start_date, storm_label, day_boundary, legend_loc in [
    (axes[0], dst_g3, kp_g3, ap_g3, time_dst_g3, time_kp_g3,
     "2024-05-01", "G3 Storm (2 May)", "2024-05-02 00:00", 'upper left'),
    (axes[1], dst_g2, kp_g2, ap_g2, time_dst_g2, time_kp_g2,
     "2024-05-05", "G2 Storm (6 May)", "2024-05-06 00:00", 'upper right'),
]:
    line1 = ax1.plot(time_dst, dst_data, color='crimson', linewidth=2.5, label='Dst (nT)')
    ax1.axhline(0, color='gray', linestyle='--', linewidth=0.8)
    ax1.set_ylabel('Dst (nT)', fontsize=my_font_size - 4)
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.tick_params(axis='both', labelsize=my_font_size - 4)
    ax1.axvline(pd.Timestamp(day_boundary), color='black', linestyle='-', linewidth=1)

    ax2 = ax1.twinx()
    line2 = ax2.plot(time_kp, kp_data, color='royalblue', marker='s', linestyle='--',
                     linewidth=2.0, label='Kp')
    ax2.set_ylabel('Kp', fontsize=my_font_size - 4)
    ax2.set_ylim(0, 9)
    ax2.set_yticks([0, 3, 6, 9])
    ax2.tick_params(axis='both', labelsize=my_font_size - 4)

    ax3 = ax1.twinx()
    ax3.spines['right'].set_position(('outward', 55))
    line3 = ax3.plot(time_kp, ap_data, color='darkorange', marker='o', linewidth=2.0, label='Ap')
    ax3.set_ylabel('Ap', fontsize=my_font_size - 4)
    ax3.tick_params(axis='both', labelsize=my_font_size - 4)

    tick_locs = pd.date_range(start=start_date, periods=5, freq="12h")
    ax1.set_xticks(tick_locs)
    ax1.set_xticklabels([dt.strftime('%d %b\n%H:%M') for dt in tick_locs], fontsize=my_font_size - 6)
    ax1.set_xlabel('Time (UTC)', fontsize=my_font_size - 4)

    ax1.set_title(storm_label, fontsize=my_font_size - 2, fontweight='bold')

    lines = line1 + line2 + line3
    labels = [l.get_label() for l in lines]
    ax1.legend(lines, labels, loc=legend_loc, fontsize=my_font_size - 6)

plt.tight_layout()
plt.savefig(GEOMAGNETIC_INDICES_DIR / "geomag_combined_G3_G2_may2024.png",
            format="png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: geomag_combined_G3_G2_may2024.png")
```
